## Pandas Time Series 연습문제

이 노트북은 Pandas의 시계열 데이터를 사용하기 위한 연습문제로 구성되어 있습니다.

### 필요한 모듈 임포트

```python
# 필요한 모듈을 임포트합니다.
import numpy as np
import pandas as pd
```

```python
# pandas 버전을 확인합니다.
# [참고] 이 노트북은 pandas 2.3.2 버전에서 작성되었습니다.
pd.__version__
```

### 가상의 시계열 데이터 불러오기

```python
# 가상의 시게열 데이터를 읽고 데이터프레임을 생성합니다.
df = pd.read_csv(filepath_or_buffer = 'https://bit.ly/Time_Series_Dataset')
```

```python
# df의 처음 5행을 확인합니다.
df.head()
```

```python
# df의 정보를 확인합니다.
df.info()
```

### 연습 문제 1: 날짜시간 데이터로 변환

```python
# df에서 date를 날짜시간형으로 변환
df['date'] = pd.to_datetime(arg=df['date'], format='%Y년 %m월 %d일')
```

```python
# df의 열별 자료형을 확인하세요.
# [참고] date의 원소 자료형은 'datetime64[ns]'이어야 합니다.
df.info()
```

```python
# df에서 date를 인덱스로 설정하고, 인덱스를 오름차순 정렬하여 df에 재할당하세요.
df = df.set_index(keys='date').sort_index()
```

```python
# df의 처음 5행을 확인하세요.
df.head()
```

### 연습 문제 2: 시계열 데이터의 인덱싱 및 슬라이싱

```python
# df에서 인덱스가 '2026-02-02'인 행을 선택하세요.(인덱싱)
df.loc['2026-02-02', :]
```

```python
# df에서 인덱스가 '2026-03-01'부터 '2026-03-10'까지인 행을 잘라내세요.(슬라이싱)
df.loc['2026-03-01':'2026-03-10', :]
```

```python
# df에서 인덱스가 '2026-02-27'과 '2026-03-27'인 행을 선택하세요.(팬시 인덱싱)
df.loc[['2026-02-27', '2026-03-27'], :]
```

```python
# df에서 인덱스가 매월 10일인 행을 선택하세요.(불리언 인덱싱)
mask = df.index.day == 10
df.loc[mask]
```

### 연습 문제 3: 중복 시점 확인

```python
# df에서 인덱스가 중복이면 True, 중복이 아니면 False인 dup_mask를 생성하세요.
dup_mask = df.index.duplicated(False)
```

```python
# df에서 dup_maks가 True인 행을 선택하고, dup_rows에 할당하세요.
dup_rows = df.loc[dup_mask]
```

```python
# dup_rows를 확인하세요.
dup_rows
```

```python
# df에서 인덱스로 그룹을 설정하고, sales의 마지막 원소를 선택하여 df에 재할당하세요.
df = pd.DataFrame(df.groupby(df.index)['sales'].last())
```

### 연습 문제 4: 누락 시점 확인

```python
# df의 인덱스 최솟값과 최댓값으로 일별 기대 인덱스를 생성하고 full_idx에 할당하세요.
full_idx = pd.date_range(df.index.min(), df.index.max())
```

```python
# full_idx와 비교하여 df의 인덱스에서 누락된 인덱스를 생성하고 miss_idx에 할당하세요.
miss_idx = full_idx.difference(df.index)
```

```python
# miss_idx를 확인하세요.
miss_idx
```

```python
# df에서 누락된 인덱스를 추가하세요.
df = df.reindex(index=full_idx)
```

### 연습 문제 5: 결측값 처리

```python
# df에서 sales가 결측인 행을 확인하세요.
# [참고] df에서 sales가 결측인 행은 모두 10행입니다.
mask = df['sales'].isna()
df.loc[mask]
```

```python
# df에서 sales의 결측값을 선형보간법으로 채우고, sales에 재할당하세요.
df['sales'] = df['sales'].interpolate(method='linear')
```

```python
# df에서 sales의 평균을 확인하세요.
# [참고] 실행 결과를 반올림하여 소수점 둘째자리까지 남기면 59.40이어야 합니다.
print(f'{df['sales'].mean().round(2):.2f}')
```

### 연습 문제 6: 시차 변수 생성

```python
# df에서 sales를 1칸 아래로 이동시키고 df에 lag_1을 추가하세요.
df['lag_1'] = df['sales'].shift(1)
```

```python
# df에서 sales를 3칸 아래로 이동시키고 df에 lag_3을 추가하세요.
df['lag_3'] = df['sales'].shift(3)
```

```python
# df에서 sales를 7칸 아래로 이동시키고 df에 lag_7을 추가하세요.
df['lag_7'] = df['sales'].shift(7)
```

```python
# df의 상관계수 행렬을 확인하세요.
# [참고] 실행 결과를 반올림하여 소수점 둘째자리까지 남기면 sales와 lag_7의 상관계수는 0.88이어야 합니다.
df.corr().round(2)
```

```python
# df에서 sales만 남기고 df에 재할당합니다.
df = df.filter(like='sales')
```

### 연습 문제 7: 이동 평균 계산

```python
# df에서 sales의 7일 이동 평균을 확인하세요.
# [참고] 최소 구간을 3으로 설정하세요.
df['sales'].rolling(window=7, min_periods=3).mean().head(10)
```

```python
# df에서 sales의 처음부터 누적된 평균을 계산합니다.
# [참고] 실행 결과, 마지막 원소의 인덱스는 2026-03-31, 값은 59.404889이어야 합니다.
df.expanding().mean().tail(1)
```

### 연습 문제 8: 변화량 기반 변수 생성

```python
# df에서 sales의 전일 대비 차분을 계산하세요.
# [참고] 실행 결과, 첫 번째 원소는 NaN이고 두 번째 원소는 -3.14이어야 합니다.
df['sales'].diff(1).head()
```

```python
# df에서 sales의 전일 대비 증감률을 계산하세요.
# [참고] 실행 결과, 첫 번째 원소는 NaN이고 두 번째 원소는 -0.065254이어야 합니다.
df.pct_change(1).head()
```

### 연습 문제 9: 시간 단위 재구성

```python
# df에서 토요일이 끝나는 주간 구간으로 sales의 평균을 계산하세요.
# [참고] 실행 결과, 첫 번째 원소의 인덱스는 2026-01-03, 값은 56.886667이어야 합니다.
df.resample(rule='W-SAT')['sales'].mean().head()
```

```python
# df에서 토요일이 끝나는 주간 구간으로 sales의 평균을 계산하고, 전주 대비 증감률을 확인하세요.
# [참고] 실행 결과, 마지막 원소의 인덱스는 2026-04-04, 값은 0.027981이어야 합니다.
df.resample(rule='W-SAT')['sales'].mean().pct_change(1).tail(1)
```

```python
# df에서 sales의 인덱스가 매주 월요일인 행을 선택하고, 전체 평균을 계산하세요.
# [참고] 실행 결과, 값은 48.802564이어야 합니다.
df.asfreq(freq='W-MON').mean()
```
